Napisz definicję funkcji HowManyIntegers(N), która zwróci jak wiele liczb z zakresu [1,N) składa się wyłącznie z cyfr {0,2,7,9}. <br><br>
Czyli np. <br>
HowManyIntegers(1) = 0 <br>
HowManyIntegers(10) = 3 # 2,7,9 <br>
HowManyIntegers(28) = 6 # 2,7,9,20,22,27 <br>

Twoje rozwiązanie powinno mieć jak najmniej niebiałych znaków. Czy uda się w 69? A może da się jeszcze mniej?

````python
import re
def HowManyIntegers(n):
    return sum(1 for x in range(n) if re.match('^[0279]+$',str(x)))-1

print(HowManyIntegers(1) == 0) 
print(HowManyIntegers(10) == 3) # 2,7,9 
print(HowManyIntegers(28) == 6) # 2,7,9,20,22,27
````

Gra słów, ale warto się zastanowić: **dlaczego wynik to `(True, True, True)`?** Odpowiedź znajdziesz pod komórką.

````python
# A poem
love = this

this is love, love is not True or False, love is love
````

* `love = this` – obie nazwy wskazują na **ten sam obiekt** (moduł `this`), więc `this is love` to `True`. Operator `is` porównuje tożsamość obiektów, a nie ich wartość (do tego służy `==`).
* `love is not True or False` czyta się jako `(love is not True) or False`, bo `is not` wiąże silniej niż `or`. Moduł nie jest obiektem `True`, więc wynik to `True`.
* `love is love` – każdy obiekt jest tożsamy z samym sobą.
* Trzy wyrażenia oddzielone przecinkami tworzą krotkę, dlatego Jupyter wyświetla `(True, True, True)`.
