### Przypomnienie

* `lambda x: wyrażenie` – anonimowa, jednowyrażeniowa funkcja,
* `map(func, iterable)` – stosuje `func` do każdego elementu,
* `filter(func, iterable)` – zostawia elementy, dla których `func` zwraca wartość prawdziwą,
* `functools.reduce(func, iterable, initial)` – „zwija” elementy do jednej wartości: `func(func(func(initial, x1), x2), x3)…`; `initial` jest opcjonalne i zwracane dla pustego `iterable`.

`map` i `filter` są **leniwe** – zwracają iterator, który liczy kolejne elementy dopiero wtedy, gdy ktoś po nie sięgnie.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

squares = map(lambda x: x ** 2, numbers)
evens = filter(lambda x: x % 2 == 0, numbers)
total = reduce(lambda acc, x: acc + x, numbers, 0)  # ((((0 + 1) + 2) + 3) + 4) + 5

print(squares)        # <map object ...> – nic jeszcze nie policzono
print(list(squares))  # dopiero teraz
print(list(evens))
print(total)
````

</details>

<details>
<summary><b>Ciekawostka:</b> nieskończone ciągi</summary>

Dzięki leniwości można pracować na ciągach, które nigdy się nie kończą – trzeba tylko wziąć z nich tyle elementów, ile potrzeba. `itertools.count()` liczy 0, 1, 2, … w nieskończoność, a `itertools.islice` bierze pierwsze `n` elementów:

```python

squares = map(lambda x: x ** 2, count())  # nieskończony ciąg kwadratów
print(list(islice(squares, 5)))
```
```
[0, 1, 4, 9, 16]
```

`list(squares)` nigdy by się nie skończyło – a `map` nic nie liczy, dopóki `islice` nie poprosi o kolejny element.
